# FEATURE ENGNEERING

## -Feature Engineering is the process of creating, transforming, selecting, or modifying input features (columns) from raw data so that a machine learning model can learn patterns more effectively and achieve better performance.

## --Feature Engineering = Converting raw data into meaningful features that help the model make better predictions

## FEATURE CREATION

### --CREATING NEW FEATURE FROM EXISTING ONE TO MAKE A BETTER PREDICTION

### --ONE HOT ENCODING COMESN UNDER THIS

### EXAMPLE

### --IF YOU HAVE THE COLOUMN'S WEIGTH AND HEIGTH , IT WOULD BE GOOD IF YOU CREATE A NEW COLOUMN "BMI"

## FEATURE TRANSFORMATION

### --CHANGING THE FEATURE VALUE'S TO A BETTER VALUE

### --LABEL ENCODING,DATA NORMALISATION,DATA STANDARDISATION COMES UNDER THIS 

### --IF YOU HAVE CATEGORICAL VALUES , DO LABEL ENCODING 

### -- IF YOU HAVE THE VALUES OF THE DIFFRENT RANGE IN UNMATCHABLE RANGE THEN DO DATA NORMALISATION

## FEATURE EXTARCTION

### --USEFULL INFORMATION ARE EXTRACED FROM THE FEATURES AND CREATED AS A NEW FEATURE

### --IT IS USED WHEN THE RAW DATA CANNOT BE USED DIRECTLY 

### EXAMPLE

### WHEN YOU HAVE THE DATA LIKE THIS 2026-06-24

### EXTRACT THIS AS YEAR MONTH DAY AND CREATE A NEW FEATURE SEPERATLY

## FEATURE SELECTION

### SELECTING ONLY THE IMPORTANT FEATURE

### REMOVE THE FEATURES WHICH DO NOT HELP IN PREDICTING

### REMOVE THE FEATURE WHICH HAVE MORE MISSING VALUES

### EXAMPLE

### IN THE CASE OF HOUSE PRICE PREDICTION , THE HOUSE ID IS UNWANTED THIS MAKES CHANCE OF OVERFITTING SO REMOVE IT 

## LOG TRANSFORMATION

### -- IT COMES UNDER FEATURE TRANSFORMATION

### RIGTH SKEWED DATA --> MOST OF THE VALUES ARE SMALL WHILE FEW ARE EXTREMELY LARGE 

### --SOME ALGORITHM PERFORM EXTREMLY POOR WHEN THE DATA IS RIGTH SKEWED

### WE APPLY THE LOG FUNCTION TO ALL THE DATA IN THE FEATURE AND REDUCE ITS RANGE , THE GAPS BETWEEN THE DATA BECOME LOW 

### THEN THE DATA BECOMES MORE SUITABLE FOR THE MODEL 

### --Log Transform compresses large values and reduces right skewness.

## POLYNOMIAL FEATURES

### IT COMES UNDER FEATURE TRABSFORMATION 

### --USED WHEN THE RELATIONSHIP AND THE TARGET IS NOT LINEAR (STRAIGTH)

### -We use Polynomial Features when the relationship between input features and target is non-linear. They create higher-order terms such as x² and x³, allowing simple models like Linear Regression to learn complex patterns.

### --– POLYNOMIAL FEATURES HELP THE MODEL LEARN CURVED RELATIONSHIPS
