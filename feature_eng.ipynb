# FEATURE ENGNEERING

## -Feature Engineering is the process of creating, transforming, selecting, or modifying input features (columns) from raw data so that a machine learning model can learn patterns more effectively and achieve better performance.

## --Feature Engineering = Converting raw data into meaningful features that help the model make better predictions

## FEATURE SELECTION

### REMOVE THE FEATURE WHICH HAVE MORE MISSING VALUES

## LOG TRANSFORMATION

### -- IT COMES UNDER FEATURE TRANSFORMATION

### RIGTH SKEWED DATA --> MOST OF THE VALUES ARE SMALL WHILE FEW ARE EXTREMELY LARGE 

### --SOME ALGORITHM PERFORM EXTREMLY POOR WHEN THE DATA IS RIGTH SKEWED

## POLYNOMIAL FEATURES

### IT COMES UNDER FEATURE TRANSFORMATION 

### --USED WHEN THE RELATIONSHIP AND THE TARGET IS NOT LINEAR (STRAIGTH)

### -We use Polynomial Features when the relationship between input features and target is non-linear. They create higher-order terms such as x² and x³, allowing simple models like Linear Regression to learn complex patterns.

### IF WE HAVE FEATURES EXPERIRENCE AND SALARY, WE CREATE EXPRENICE^2 , SALARY^2 AND EXPRENICE^SALARY

## Polynomial Features are new features created by raising existing numerical features to a power or by multiplying them together, so the model can learn non-linear relationships.

### --– POLYNOMIAL FEATURES HELP THE MODEL LEARN CURVED RELATIONSHIPS

### WHEN SHOULD WE USE THIS 

### --IF THE DATA IS NOT LINEAR 

### --YOU ARE USING A LINEAR MODEL 

### --WHEN YOU HAVE ONLY FEW NUMERICAL COLOUMNS

## INTERACTION FEATURES

### --Interaction Features are new features created by multiplying two or more existing features to capture their combined effect on the target variable.

### --NEW FEATURES CREATED BY MULTIPLTING TWO OR MORE FEATURES

### --COMES UNDER FEATURE CREATION

### EXAMPLE

### --IN OREDER TO PREDICT THE SALARY  WE GIVE SKILL_SCORE  AND EXPRECIENCE SEPATELY

### --HERE WE CAN MAKE A NEW FEATURE BY MULTIPLYING SKILL_SCORE AND EXP TO GET A BETTER FEATURE

### --USUALLY WE DONT DROP THE RESPECTIVE FEATURE WHICH WE USED TO CREATE A NEW , WE KEEP ALL THREE FEATURES -- THAT IS DECIDED IN FEATURE SELECTION

## RATIO FEATURES

### --THE FEATURES WHICH ARE MADE BY DIVIDING TWO FEATURES

## --Ratio Features are new features created by dividing one feature by another to capture their relative relationship.

## --Sometimes the relationship between two features is more important than the individual features.

### IN THIS CASE WE USE IT

### --COMES UNDER FEATURE CREATION

### EXAMPLE

### -- TAKE HOUSE PRICE PREDICTION

### --THERE ARE TWO FEATURE AREA , BEDROOM

### -- IF WE CREATE NEW FEATURE AREA PER BEDROOM , IT MIGTH BE VERY GOOD FEATURE 

### --SO FOR AREA PER BEDROOM WE DIVIDE ONE BY ANOTHER , THIS IS CALLED RATIO FEATURES
