```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

### LOADING DATA

```python
df=pd.read_csv('online_retail.csv')
df.head()
```

```python
df.info()
```

### we have loaded and checked  , we have got an info about the dataset, while it seems there may null values in Customer_id

### -- before handling null we may eliminate duplicate rows

## CHECKING DUPLIACTE ROWS

```python
df.duplicated().sum()
```

### -- it means there are 5268 duplicate rows

### -- then we should delete the duplicated rows by only keeping the first row only 

```python
df.drop_duplicates(keep='first', inplace=True)
```

#### -- here we have used drop_dupliactes to remove the duplicate row , keep='first' -- dentoes to keep one row from the duplicate ,i,e store one row form duplicated,, inplace=true denote do it in inplace

### -now we again recheck using df.duplicated.sum()

```python
df.duplicated().sum()
```

### now we have ZERO duplicated rows

## NULL VALUES 

```python
df.isnull().sum()
```

### there are many null values here

### --- CUSTOMER ID HAVE 13K NULL VALUES AND THAT CLOUMN HAVE NO CONNECTING WITH TARGET ,SO DROPING IT IS THE BEST APPROACH

```python
df.drop(columns=['CustomerID'], inplace=True)
```

### --even DESCRIPTON HAVE MORE NULL VALUE AND NO CONNECTION WITH TARGET , SOO DROPING IS THE BEST APPROACH

```python
df.drop(columns=['Description'], inplace=True)
```

### WE HAVE DROPED THE NULL VALUE , ONCE WE MIGTH CHECK IS THERE ANY NULL VALUE AGAIN

```python
df.isnull().sum()
```

### THEREFORE NO NULL VALUES , WE HAVE HANDLED NULL VALUES
