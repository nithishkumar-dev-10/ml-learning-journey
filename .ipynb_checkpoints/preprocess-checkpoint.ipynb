## DATA NORMALIZATION

### Data normalization is the process of scaling numerical features to a common range (usually 0–1) to improve the performance and stability of machine learning algorithms.

### we have an predefined function to do those

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
df=pd.read_csv("train.csv")
scaler=MinMaxScaler()
df['normalised_fare']=scaler.fit_transform(df[['Fare']])
df.head()
```

### STEP-1 --- IMPORT PANDAS , SKLEARN AND MINMAXSCALER

### STEP -- 3 ---- CREATE A VARIABLE CALLED SCALER WHICH HOLDES MINMAXSCALER() OBJECT 

### STEP 4 ---- CREATE A NEW COULMN WHILE THE RHS == scaler.fit_transform("coloum to be normalized ")

### IN STEP 3 WE ARE CREATING A OBJECT "MINMAXSCALER" IN A VARIABLE CALLED "SCALER"

### WHILE IN STEP FOUR WE ARE TRANFORMING IT BY USING THE SYNTAX  scaler.fit_transform("coloum-to-be-transformed")  side by we are storing in a new coloumn



## LABEL ENCODING 

### Label Encoding is a data preprocessing technique used in machine learning to convert categorical (text) data into numerical labels so that algorithms can process the data.

### COVERTS THE CATEGORICAL COLOUMNS INTO NUMERICAL VALUES BY GIVING A NUMERCIAL VALUES FOR EACH UNIQUE ELEMENT,IF THE ELEMENT IS REPEATED THE SAME NUMBER IS ASSINGNED ONCE AGAIN

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
df=pd.read_csv("train.csv")
scaler=LabelEncoder()
df['Sex']=scaler.fit_transform(df[['Sex']])
df.head()
```

### HERE WE ARE COVERTING THE COLOUMN TO NUMERICAL VALUE ,BY DEAFULT THE GENDER MALE TAKEN AS 1,WHILE FEMALE TAKEN AS 0

### LABLING IN A MANUAL WAY , I,E CUSTOMISING THE LABEL 

```python
import pandas as pd
df=pd.read_csv("train.csv")
df['Sex']=df['Sex'].map({"male":1,"female":0})
df.head()
```

### FOR MANUAL ENCODING WE USE MAP FUNCTION HERE

### df['coloumn'].map({condition}) --- is the proper syntax

## ONE - HOT ENCODING

```python
import pandas as pd
df=pd.read_csv("train.csv")
df=pd.get_dummies(df,columns=["Sex"])
df.head()

```
