# LOGISTIC REGRESSION

## Logistic Regression is a supervised machine learning algorithm used for classification problems, where the target variable is categorical (usually 0 or 1).

### -- RATHER THAN IT'S NAME OF THE REGRESSION IT IS A CLASSIFICATION MODEL 

### -- WE USE THIS TO PREDICT THE CATEGORIAL TARGET VATRIABLE LIKE ->(YES,NO) AND ETC ..

## ----IT FITS A "S"-SHAPED  LOGISTICS FUNCTION

## --THE "S" - SHAPED CURVE GOES FROM 0 - 1 , WHICH IS USED TO PREDICT YES OR NO 

### --IF WE GET THE PROBABLITY LESS THAN 0.5 (50%) THEN NO IF IT IS GREATER THAN 0.5 (50%) THEN WE MIGTH GET AS YES --> IT IS ACCORDING TO OUR PROBLEM STATEMENT 

### WE CAN HAVE ONE OR MORE INPUT COLOUMN'S FOR PERDICTING A TARGET VALUE

### IT DOESNT HAVE THE CONCPET OF RESIDUAL , INSTEAD WE HAVE THE CONCEPT OF MAXIMUM LIKELIHOOD

## ODDS AND LOGS

##  AS THE LOGISTIC REGRESSION IS A LINEAR MODEL IT RANGES FROM -INFINTE TO INFINTE BUT THE OUTPUT PROBALITY RANGES TO 0-1 ,SO WE CONVERT THE PROBALITY TO ODDS AND LOG(ODDS)

## --ODDS REPRESENTS THE RATION OF PROBABILITY OF THE EVENT OCUURS TO THE PROBABILITY OF THE EVENT DOESNT OCCURS

### ODDS = P(E)/P(E'). <------OR------>.  0DDS = P/(1-P)   

### ---WHY IT MATTERS


### -- THIS SAY HOW MUCH LIKELY A EVENT IS TO BE HAPPEN THAN EVENT DOESNT HAPPEN


### --THE ODDS RANGES FROM 0 - INFINTE SO WE CONVERT IT INTO LOG-ODDS

## LOG-ODDS

### LOGRATHIMIC OF THE ODDS

### ---LOG(P/(1-P))

### --- LOG(P(E)/P(E'))

### -- THIS IS USED TO CONVERT THE RANGE FROM 0-INFINTE TO - INFINITE TO INFINTE

### --AS THE ODDS RANGES ONLY FROM 0-INFINTE WE USE LOG-ODDS TO CONVERT ITS RANGE --> -INFINTE TO INFINTE

## WHY ODDS AND LOG ODDS MATTER

### -- THE LOGISTIC REGRESION MODEL IS A LINEAR MODEL WHICH MEAN IT RANGES FROM -INFINTE TO INFINITE

### --BUT THE PROBABLITY ONLY RANGES FROM 0-1 

### -- SO WE NEED TO CONVERT THE RANGE OF PROBABLITY -INFINTE TO INFINITE

### --FIRST WE CONVERT TO ODDS -->  P(E)/P(E') --->P/(1-P)

### --NOW WE GET THE RANGE FROM 0 TO INFINTE

### -- NOW TAKE THE LOG FOR THE ODDS WE GET LOG-ODDS

### --LOG(ODDS)--->LOG(P(E)/P(E'))----->LOG(P/(1-P))

### --THIS GIVE'S THE RANGE OF -INFINTE TO INFINITE

## SIGMOID FUNCTION

### --The Sigmoid Function is a mathematical function that converts any real number (−∞ to +∞) into a probability value between 0 and 1.

### --FOR THE LINEAR MODEL WE GET THE ODDS AND LOD(ODDS)

### ---> HOW DO WE CONVERT BACK TO THE PROBABLITY

### ---> WE USE THE SIGMOID FUNCTION TO CONVERT THE PREDICTION OF THE LINEAR MODEL INTO THE RANGE OF 0-1 

### --AS THE LOGISTIC REGRESION IS A LINEAR MODEL WE GET THE OUTPUT IN THE RANGE OF -INFINTE TO INFINTE

### -- SO TO CONVERT IT INTO A PROBABLITY RANGE (0-1) WE USE SIGMOID FUNCTION 

### sigma(z)=1/(1+(e^-z))  --- THIS IS THE MATHEMATICAL REPRESENTATION

# DECISION TREE

### -IT ASK QUESTION AND CATEGORISE IT WITH YES OR NO 

### --FIRST IT ASK A QUESTION THEN CATEGORISE LIKE WISE ACCORDING TO TH EPROBLEM STATEMENT IT ASKS QUESTION'S AND CATEGORISE N TIMES

### --IT ASK QUESTION AND CLASSIFIES UNTILL A POINT WHERE NO MORE QUESTION CAN BE ASKED 


## NODES

### --THE FIRST QUESTION OR THE FIRST(TOP) NODE IS KNOWN AS ROOT NODE

### --THE REMAINING NODES ARE KNOW AS INTERNAL NODES AND SOMETIMES LEAF NODE'S

### INTERNAL NODE VS LEAF NODE

### --INTERNAL NODE HAVE THE ARROWS POINTING THEM AND ALSO ARROWS FROM THEM TO ANOTHER

### -WHILE THE LEAF NODE ONLY HAVE ARROW'S POIINTING THEM ---> THE LAST NODES WITHOUT  ARROW FROM THEM TO OTHER -->IT CAN EXIST MIDDLE OF THE TREE ALSO

## HOW TO MAKE THE DECISION TREE

### -FIRST TAKE THE MOST IMPORTANT FEATURE AND ASK YES OR NO AND CLASSIFIY,THEN GO WITH NEXT IMPORTANT FEATURE LIKE WISE REPEAT THE PROCESS UNTILL THE LEAST IMPORTANT FEAUTURE IS DONE AND GET THE PREDICTION

### -A Decision Tree works by finding the most useful feature, asking a Yes/No question on it, splitting the data accordingly, and repeating the process until it can make a predic

## IMPURITY

## --Impurity is the amount of mixing of different classes in a node.

### -- A NODE CAN CONTAIN BOTH THE POSITIVE AND NEGATIVE CLASS 

### --IF A NODE CONTAINS BOTH POSITIVE AND NEGATIVE CLASS THEN IT IS CONSIDERED AS IMPURE 

### EXAMPLE

### SUPPOSE A NODE CONTAINS FIVE CUSTOMER , WHILE 3 ARE ACTIVE AND 2 ARE INACTICVE ,THEN THE NODE IS CONSIDERED TO BE IMPURE 

## GINI IMPURITY

### -IT IS A METRIC USED TO MEASURE HOW IMPURE A NODE IS.

### -IT IS A METHOD THAT IS USED TO CALCULATE HOW IMPURE A NODE IS .

### -MIXING OF DIFFERNT CLASS-->IMPURITY

### -MAGNITUDE OF IMPURITY-->GINI IMPURITY

## FORMULA

## Gini = 1 - (P(Yes)² + P(No)²)

### --WHERE P(YES)-> PROBALITY OF TRUE CASE , WHILE P(NO)-->PROBABLITY OF FALSE CASE

### GINI IMPURITY RANGES FROM 0-0.5

### 0.0 ---> Perfectly Pure

### 0.1 ----> Very Pure

### 0.2 ---> MOSTLY PURE

### 0.3 ---> SLIGTHLY IMPURE

### 0.4 ---> IMPURE 

### 0.5 ---> MORE IMPURE 

### LOW GINI SCORE --> GOOD SPLIT , WHY IT MATTER'S ?

### -- SO FIND THE SPLIT WITH LOWEST GINI SCORE AND MAKE MORE SPLIT ON IT AND REPEAT THE PROCESS AND PREDICT IT 

### --DONT DIRETCLY PREDICT IN THE SPLIT WITH LOW GINI SCORE , CHOOSE THE BEST SPLIT AND REPEAT THE PROCESS AND THEN PREDICT

## ENTROPY

### Entropy is another metric used to measure the impurity (or randomness) of a node in a Decision Tree.

### SIMILAR TO GINI BUT A DIFFERENT METHOD

### FORMULA 

### Entropy = -(P(Yes)\log_2(P(Yes)) + P(No)\log_2(P(No)))

### - IT RANGES FROM 0-1 ,WHERE VALUES CLOSE TO ZERO IS PURE AND VALUES CLOSE TO ONE IS IMPURE 

### 0.0 <-> 0.4 --> PURE 

### 0.4 AND ABOVE IS IMPURE 

## GINI VS ENTROPY

### -GINI IS MORE FASTER

### FORMULA FOR GINI -->Gini = 1 - (P(Yes)² + P(No)²)

### FORMULA FOR ENTROPY --> Entropy = -(P(Yes)\log_2(P(Yes)) + P(No)\log_2(P(No)))

### RANGE OF GINI 0-0.5

### RANGES OF ENTROPY 0-1

### INTERMS OF ACCURACY --> Gini ≈ Entropy

### INTERMS OF SPEED ---> GINI WINS

## INFORMATION GAIN

### Information Gain measures how much impurity (Entropy) is reduced after a split.

### -THE DIFFRENCE BETWEEN THE IMPURITY BETWEEN THE TWO SPLIT -- PARENT NODE AND CHILD NODE , NOT TWO DIFFRENT(RIGTH AN LEFT) NODES 

### FORMULA 

### IG = Entropy(Parent) - Weighted\ Entropy(Children)


### --IT RANGES FROM 0-1

### THE VALUES CLOSE TO ZERO IS NO OR LESS IMPURITY IS REMOVED SO NOT GOOD

### THE VALUES CLOSE TO ONE IS PERFECT SPLIT AND MORE IMPURITY IS REMOVED
