# CROSS VALIDATION AND GRIDSEARCHCV

## Instead of relying on a single train-test split, which may give a biased estimate depending on the data split, we use K-Fold Cross Validation to train and evaluate each candidate model across multiple folds and compare their average CV scores to select the best model; then we use GridSearchCV, which internally uses Cross Validation, to find the best hyperparameter combination for that selected model.

## REGRESSION MODELS

## MSE (MEAN SQUARE ERROR)

### --THE AVERAGE OF THE SUM OF SQUARE OF THE RESIDUAL

### -- THE AVERAGE OF THE SUM OF SQUARE OF THE DIFFERENECE BETWEEN THE ACTUAL VALUE AND THE PREDICTED VALUE

### --IT TELLS ON AVERAGE HOW MUCH MY PREDICTION IS WRONG 

### LOW MSE ---> GOOD MODEL , HIGH MSE --> BAD MODEL 

## RMSE - (ROOT MEAN SQUARE ERROR)

### -- IT IS JUST THE ROOT OF MSE

### --WHY ROOT?

### --AS MSE VALUES ARE HIGH THE ROOT BRINGS BACK TO ITS ORG SCALE 

## MAE (MEAN ABSOLUTE ERROR)

### THE AVERAGE OF SUM OF RESIDUALS

### --LOW VALUE --> GOOD MODEL

### ---HIGH VALUE --> BAD MODEL

## MAE VS MSE

### IN MAE WE CALCULATE THE AVERAGE OF SUM OF RESIDUALS WHILE IN MSE WE CALCULATE THE AVERAGE OF SUM OF SQUARED RESIDUALS

## R SQUARE

### --IT SAYS HOW MUCH VARIATION IN THE TARGET VARIABLE IS EXPLAINED BY THE MODEL

### --IT IS ALREADY WRITTEN IN ML_MODLES.IPYNB

## CLASSIFICATION MODELS

## ACCURACY

### --HOW MANY PREDICTION IS CORRECT FROM THE TOTAL PREDICTION

### --ACCURACY = CORRECT PREDICTION/TOTAL PREDICTION

### --RANGES FROM 0-1 , WHERE THE NUMBERS CLOSETS TO 1 IS GOOD MODEL WHILE CLOSEST TO 0 IS BAD MODEL

## PRECISION

### OUT OF ALL SAMPELS THAT MADE PREDICTION AS POSITVE ,HOW MANY SAMPLE WERE POSITVE 

### PRECISION= TP/TP+FP

### TP--> TRUE POSITIVE,FP--> FALSE POSITIVE


### --RANGES FROM 0-1 , WHERE THE NUMBERS CLOSETS TO 1 IS GOOD MODEL WHILE CLOSEST TO 0 IS BAD MODEL

### ACCURACY VS PRECISION

### ACCURACY TAKES ALL PREDICTION AND CHEKS HOW MANY PREDICTION IS CORRECT (DO CARE ABOUT NEGATIVE PREDICTION)

### WHILE PRECISION TAKES ALL POSITIVE PREDCITION AND CHECKS HOW MANY ARE ACTUACLLY POSITIVE (DO NOT CARE ABOUT NEGATIVE PPREDICTION)

## RECALL

### --OUT OF ALL POSTIVE CASE HOW MANY DID THE MODEL FIND

### RECALL = TP/TP+FN

### TP-->TRUE POSITIVE , FN-->FALSE NEGATIVE

## PRECISION VS RECALL

### PRECISION CHECKS IN OUT OF ALL PREDICTED VALUES AS POSITVE HOW MUCH ARE ACTUALLY POSITIVE

### RECALL CHECKS OUT OF ALL POSITIVE CASES HOW MUCH DID OUR MODEL FIND

## F1 SCORE

### --F1-Score is the harmonic mean of Precision and Recall.

### IF A MODEL HAVE PRECISION HIGH AND LOW RECALL , WHILE OTHER HAVE HIGH RECALL AND LOW PRECISION HOW DO YOU SAY WHICH IS THE BETTER MODEL 

### TO FIND IT WE USE THE HARMONIC MEAN THATS IS CALLED F1 SCORE

### F1 = 2 * (PRECISION*RECALL)/(PRECISION+RECALL)

### AUC IS THE AREA UNDER THE CURVE 

### THIS SAY HOW WELL THE MODEL SEPERATES THE POSITVE AND NEGATIVE VALUES
