## CROSS VALIDATION

### IT IS A TECHNIQUE IN WHICH WE EVALUTE THE DATA , ON WHICH HOW WELL IT PERFORMS ON A NEW OR UNSEEN DATA 

### IT CREATES A K FOLDS AND TRAINS , TESTS K TIMES 

### IT TRAINS AND TEST ON ALL THE DATA RATHER THAN ONLY ONE FIRST 75% TRAIN AND LAST 25% TEST 

### SOO IT IS BETTER FOR A MODEL EVALUATION

### -- IT IS USED TO FIND THE BEST MODEL AND BEST HYPERPARAMETER(TUNING)

# CROSS VALIDATION AND GRIDSEARCHCV

## Instead of relying on a single train-test split, which may give a biased estimate depending on the data split, we use K-Fold Cross Validation to train and evaluate each candidate model across multiple folds and compare their average CV scores to select the best model; then we use GridSearchCV, which internally uses Cross Validation, to find the best hyperparameter combination for that selected model.
