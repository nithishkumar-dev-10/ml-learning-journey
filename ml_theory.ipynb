## SUPERVISED VS UNSUPERVISED VS REINFORCEMENT

### 

## SUPERVISED

#### IT USES LABELED DATA TO TRAIN A MODEL

#### LABELED DATA MEANS OUPUT IS ALREADY KNOW, FIXED OUTPUT VARIABLE

## UNSUPERVISED 

#### USES UNLABELED DATA TO TRAIN A MODEL

#### UNLABELED DATA MEANS THERE IS NO FIXED OUTPUT VARIABLE

#### LEARNS FROM THE DATA, UNDERSTANDS THE PATTERN FROM THE DATA AND PREDICTS THE OUPUT 

## REINFORCEMENT 

### USES TRIAL AND ERROR METHOD TO ARRIVE THE DESIRED SOLUTION

### IF IT FALLS IT LOSES SOME POINT (PENALTY)

### IF IT POSITIVE REWARD IT GAINS SOME POINT-(REWARD)

### STATE - ACTION -REWARD 

## CROSS VALIDATION

### IT IS A TECHNIQUE IN WHICH WE EVALUTE THE DATA , ON WHICH HOW WELL IT PERFORMS ON A NEW OR UNSEEN DATA 

### IT CREATES A K FOLDS AND TRAINS , TESTS K TIMES 

### IT TRAINS AND TEST ON ALL THE DATA RATHER THAN ONLY ONE FIRST 75% TRAIN AND LAST 25% TEST 

### SOO IT IS BETTER FOR A MODEL EVALUATION
